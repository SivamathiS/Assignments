```python
destbl = pd.DataFrame()
destbl = pd.DataFrame(index=["Mean", "Median", "Mode","Q1:25%","Q2:50%","Q3:75%","99%","Q4:100%","IQR","1.5Rule", "Lessor", "Greater","Min","Max","LrOutlier", "GtrOutlier"], columns = quan)
for colName in quan:
    destbl.loc["Mean", colName] = dtset[colName].mean()
    destbl.loc["Median", colName] = dtset[colName].median()
    destbl.loc["Mode", colName] = dtset[colName].mode()[0]
    destbl.loc["Q1:25%", colName] = dtset[colName].quantile(0.25)
    destbl.loc["Q2:50%", colName] = dtset[colName].quantile(0.50)
    destbl.loc["Q3:75%", colName] = dtset[colName].quantile(0.75)
    destbl.loc["99%", colName] = dtset[colName].quantile(0.99)
    destbl.loc["Q4:100%", colName] = dtset[colName].quantile(1.0)
    destbl.loc["IQR", colName] = dtset[colName].quantile(0.75) - dtset[colName].quantile(0.25)
    destbl.loc["1.5Rule", colName] = 1.5*destbl.loc["IQR", colName]

    lessor = dtset[colName].quantile(0.25) - (1.5*destbl.loc["IQR", colName])
    greater = dtset[colName].quantile(0.75) + (1.5*destbl.loc["IQR", colName])
    
    destbl.loc["Lessor", colName] = lessor
    destbl.loc["Greater", colName] = greater
    destbl.loc["Min", colName] = dtset[colName].min()
    destbl.loc["Max", colName] = dtset[colName].max()

    destbl.loc["LrOutlier", colName] = "Yes" if dtset[colName].min() < lessor else "No"
    destbl.loc["GtrOutlier", colName] = "Yes" if dtset[colName].max() > greater else "No"
    
destbl    
```

```python
destbl = pd.DataFrame(index=["Mean", "Median", "Mode","Q1:25%","Q2:50%","Q3:75%","99%","Q4:100%","IQR","1.5Rule", "Lessor", "Greater","Min","Max","LrOutlier", "GtrOutlier","LrOutlier_Af","GtrOutlier_Af","Kurtosis", "Skewness"], columns = quan)
for colName in quan:
    destbl.loc["Mean", colName] = dtset[colName].mean()
    destbl.loc["Median", colName] = dtset[colName].median()
    destbl.loc["Mode", colName] = dtset[colName].mode()[0]
    destbl.loc["Q1:25%", colName] = dtset[colName].quantile(0.25)
    destbl.loc["Q2:50%", colName] = dtset[colName].quantile(0.50)
    destbl.loc["Q3:75%", colName] = dtset[colName].quantile(0.75)
    destbl.loc["99%", colName] = dtset[colName].quantile(0.99)
    destbl.loc["Q4:100%", colName] = dtset[colName].quantile(1.0)
    destbl.loc["IQR", colName] = dtset[colName].quantile(0.75) - dtset[colName].quantile(0.25)
    destbl.loc["1.5Rule", colName] = 1.5*destbl.loc["IQR", colName]

    lessor = dtset[colName].quantile(0.25) - (1.5*destbl.loc["IQR", colName])
    greater = dtset[colName].quantile(0.75) + (1.5*destbl.loc["IQR", colName])
    
    destbl.loc["Lessor", colName] = lessor
    destbl.loc["Greater", colName] = greater
    destbl.loc["Min", colName] = dtset[colName].min()
    destbl.loc["Max", colName] = dtset[colName].max()

    destbl.loc["LrOutlier", colName] = "Yes" if dtset[colName].min() < lessor else "No"

    destbl.loc["GtrOutlier", colName] = "Yes" if dtset[colName].max() > greater else "No"

    dtset.loc[dtset[colName] < lessor, colName] = lessor
    dtset.loc[dtset[colName] > greater, colName] = greater

    destbl.loc["LrOutlier_Af", colName] = "Yes" if dtset[colName].min() < lessor else "No"
    destbl.loc["GtrOutlier_Af", colName] = "Yes" if dtset[colName].max() > greater else "No"

    destbl.loc["Kurtosis", colName] = dtset[colName].kurtosis()
    destbl.loc["Skewness", colName] = dtset[colName].skew()

    
destbl    
```

```python
    def freqTable(dataset, columnName):
        freqTable=pd.DataFrame(columns=["Unique_Values","Frequency","Relative Frequency","Cusum"])
        freqTable["Unique_Values"]=dataset[columnName].value_counts().sort_index()
        freqTable["Frequency"]=dataset[columnName].value_counts().values
        freqTable["Relative Frequency"]=freqTable["Frequency"]/dataset[columnName].value_counts().sort_index().sum()
        freqTable["Cusum"]=freqTable["Relative Frequency"].cumsum()
        return freqTable
```
