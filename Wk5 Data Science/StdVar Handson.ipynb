```python
import pandas as pd
dtset = pd.read_csv("Placement.csv")
dtset = dtset.drop(columns=['sl_no'])

# Quantitative columns
quan = dtset.select_dtypes(include=['number']).columns

# Qualitative columns
qual = dtset.select_dtypes(include=['object', 'category']).columns
```

```python
destbl = pd.DataFrame()
destbl = pd.DataFrame(index=["Mean", "Median", "Mode","Q1:25%","Q2:50%","Q3:75%","99%","Q4:100%","IQR","1.5Rule", "Lessor", "Greater","Min","Max","kurtosis", "skew", "Std", "Var"], columns = quan)
for colName in quan:
    destbl.loc["Mean", colName] = dtset[colName].mean()
    destbl.loc["Median", colName] = dtset[colName].median()
    destbl.loc["Mode", colName] = dtset[colName].mode()[0]
    destbl.loc["Q1:25%", colName] = dtset[colName].quantile(0.25)
    destbl.loc["Q2:50%", colName] = dtset[colName].quantile(0.50)
    destbl.loc["Q3:75%", colName] = dtset[colName].quantile(0.75)
    destbl.loc["99%", colName] = dtset[colName].quantile(0.99)
    destbl.loc["Q4:100%", colName] = dtset[colName].quantile(1.0)
    destbl.loc["IQR", colName] = dtset[colName].quantile(0.75) - dtset[colName].quantile(0.25)
    destbl.loc["1.5Rule", colName] = 1.5*destbl.loc["IQR", colName]

    lessor = dtset[colName].quantile(0.25) - (1.5*destbl.loc["IQR", colName])
    greater = dtset[colName].quantile(0.75) + (1.5*destbl.loc["IQR", colName])
    
    destbl.loc["Lessor", colName] = lessor
    destbl.loc["Greater", colName] = greater
    destbl.loc["Min", colName] = dtset[colName].min()
    destbl.loc["Max", colName] = dtset[colName].max()

    destbl.loc["kurtosis", colName] = dtset[colName].kurtosis()
    destbl.loc["skew", colName] = dtset[colName].skew()
    destbl.loc["Std", colName] = dtset[colName].std()
    destbl.loc["Var", colName] = dtset[colName].var()

destbl    
```
