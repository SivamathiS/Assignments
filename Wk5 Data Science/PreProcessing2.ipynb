```python
import pandas as pd
dtset = pd.read_csv("Placement.csv")
dtset = dtset.drop(columns=['sl_no'])
dtset
```

```python
dtset.isnull().sum()
```

```python
# Quantitative columns
quan = dtset.select_dtypes(include=['number']).columns

# Qualitative columns
qual = dtset.select_dtypes(include=['object', 'category']).columns

dtset[quan]
```

```python
import numpy as np
from sklearn.impute import SimpleImputer
imp = SimpleImputer(missing_values=np.nan, strategy = 'mean')
imp.fit(dtset[quan])
df = imp.transform(dtset[quan])
df = pd.DataFrame(df, columns = quan)
df
```

```python
catset = dtset[qual]
catset
```

```python
mergeset = [df,catset]
preprocessed = pd.concat(mergeset, axis = 1)
preprocessed
```

```python
preprocessed.isnull().sum()
```

```python
preprocessed.to_csv("processedPlacementData.csv", index=False)
```
