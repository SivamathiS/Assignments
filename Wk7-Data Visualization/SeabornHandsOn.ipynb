```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

dtset = pd.read_csv("Placement.csv")
dtset = dtset.drop(columns=['sl_no'])
print("No of rows:", dtset.shape[0])
print("No of columns:", dtset.shape[1])
print ("Total records:", dtset.shape)
```

```python

updset = dtset.copy()
updset['salary'] = updset["salary"].fillna(updset["salary"].mean())
updset.fillna(0)

quan = updset.select_dtypes(include=['number']).columns
quan
```

```python
import seaborn as sb
sb.histplot(
    data=updset,
    x='ssc_p',
    kde=True,
    line_kws={'color': 'red'}
)
sb.kdeplot(data=updset, x='ssc_p', color='green')

plt.show()
```
