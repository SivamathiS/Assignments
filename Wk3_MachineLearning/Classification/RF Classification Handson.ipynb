```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
dtset = pd.read_csv("Social_Network_Ads.csv")
dtset.columns
```

```python
dtset = pd.get_dummies(dtset,drop_first=True)
independant = dtset.drop('User ID', axis=1)
dept = dtset[['User ID']]
```

```python
dtset
```

```python

from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(independant, dept, test_size = 1/3, random_state=0)
```

```python
from sklearn.ensemble import RandomForestClassifier
classifier = RandomForestClassifier(n_estimators = 10, criterion = 'entropy', random_state = 0)
classifier.fit(x_train, y_train)
```

```python
y_predict = classifier.predict(x_test)
from sklearn.metrics import r2_score
rscore=r2_score(y_test, y_predict)
rscore
```

```python
y_pred = classifier.predict(x_test)
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)     
cm
```

```python
from sklearn.metrics import classification_report
clf_report = classification_report(y_test, y_pred)
print(clf_report)
```
