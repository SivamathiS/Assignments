```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
dtset = pd.read_csv("50_Startups.csv")
dtset.columns
```

```python
dtset = pd.get_dummies(dtset,drop_first=True)
independant = dtset.drop('Profit', axis=1)
dept = dtset[['Profit']]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(independant, dept, test_size = 1/3, random_state=0)
x_train.shape
```

```python
y_train.shape
```

```python
from sklearn.tree import DecisionTreeRegressor
regressor = DecisionTreeRegressor()
regressor = regressor.fit(x_train, y_train)
```

```python
y_predict = regressor.predict(x_test)
from sklearn.metrics import r2_score
rscore=r2_score(y_test, y_predict)
rscore
```

```python
regressor.feature_importances_
regressor.max_features_
regressor.tree_
```

```python
from sklearn import tree
tree.plot_tree(regressor)
plt.show()
```
